## Context & Methods
Research observed 5 October 2026. Public official prices only; taxes excluded. Annual equivalents divide the invoice by twelve. Five users includes the owner. No foreign-exchange conversion, product trial or quality benchmark.

### Key Assumptions

## Results
Amounts are in the currency column. Effective monthly cost requires the annual contract.


```python
def money(value):
    return str(value.quantize(Decimal('0.01'), rounding=ROUND_HALF_UP))

headers = ['Vendor / plan', 'Users', 'Networks', 'Currency', 'Monthly billing', 'Annual invoice', 'Annual effective']
table_rows = []
for record in records:
    annual = Decimal(record['annual'])
    monthly = 'Not verified' if record['monthly'] is None else money(Decimal(record['monthly']))
    table_rows.append([record['vendor'] + ' / ' + record['plan'], str(record['users']), str(record['networks']), record['currency'], monthly, money(annual), money(annual / Decimal(12))])
table_html = '<table><thead><tr>' + ''.join('<th>' + html.escape(value) + '</th>' for value in headers) + '</tr></thead><tbody>'
for row in table_rows:
    table_html += '<tr>' + ''.join('<td>' + html.escape(value) + '</td>' for value in row) + '</tr>'
table_html += '</tbody></table>'
```

```python
assert len(records) == 16
assert len({row['vendor'] for row in records}) == 8
assert all(row['users'] in (1, 5) for row in records)
assert all(row['networks'] == (4 if row['vendor'] == 'Later' else 5) for row in records)
assert all(Decimal(row['annual']) > 0 for row in records)
assert sum(row['monthly'] is None for row in records) == 4
assert {row['currency'] for row in records} == {'USD', 'GBP'}
assert 5 * 6 == 30 and 5 * 12 == 60
assert 5 * 60 == 300 and 5 * 120 == 600
assert Decimal('47') * Decimal('0.8') * 12 == Decimal('451.20')
assert 25 + 10 == 35 and 20 * 12 + 120 == 360
assert 67 + 10 == 77 and 53 * 12 + 120 == 756
assert 59 * 10 == 590 and 590 + 140 == 730
assert 49 + 4 * 10 == 89 and 490 + 4 * 100 == 890
assert 99 * 5 == 495 and 79 * 12 * 5 == 4740
assert 89 * 12 * 5 == 5340 and 349 * 12 * 5 == 20940
assert 450 + 3 * 45 == 585
assert money(Decimal('590') / 12) == '49.17'
assert money(Decimal('890') / 12) == '74.17'
print('All 19 arithmetic and scope assertions passed.')
```
